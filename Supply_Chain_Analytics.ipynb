```python

import pandas as pd

df = pd.read_csv(
    "DataCoSupplyChainDataset.csv",
    encoding="latin1"
)

print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])

print("\nColumn Names:")
print(df.columns.tolist())

print("\nFirst 5 Rows:")
display(df.head())
```

```python
print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])
```

```python
selected_columns = [
    "Order Id",
    "order date (DateOrders)",
    "shipping date (DateOrders)",
    "Days for shipping (real)",
    "Days for shipment (scheduled)",
    "Delivery Status",
    "Shipping Mode",
    "Category Name"
]

df_50k = df[selected_columns].sample(
    n=50000,
    random_state=42
).copy()

print("Selected Rows:", df_50k.shape[0])
print("Selected Columns:", df_50k.shape[1])

display(df_50k.head())
```

```python

df_50k.to_csv(
    "supply_chain_50000_rows_8_columns.csv",
    index=False
)

from google.colab import files

files.download("supply_chain_50000_rows_8_columns.csv")

```

```python

# Check missing values in each column

print("Missing Values:")
print(df_50k.isnull().sum())

print("\nTotal Missing Values:")
print(df_50k.isnull().sum().sum())

```

```python

# Check duplicate records

print("Duplicate Rows:", df_50k.duplicated().sum())

# Check unique order IDs
print("Unique Order IDs:", df_50k["Order Id"].nunique())

```

```python

date_columns = [
    "order date (DateOrders)",
    "shipping date (DateOrders)"
]

for col in date_columns:
    df_50k[col] = pd.to_datetime(
        df_50k[col],
        errors="coerce"
    )

print(df_50k[date_columns].dtypes)

print("\nInvalid or Missing Dates:")
print(df_50k[date_columns].isnull().sum())

```

```python

# Check data types and invalid duration values

duration_columns = [
    "Days for shipping (real)",
    "Days for shipment (scheduled)"
]

print(df_50k[duration_columns].describe())

print("\nMissing Duration Values:")
print(df_50k[duration_columns].isnull().sum())

print("\nZero or Negative Duration Records:")
print(
    (df_50k[duration_columns] <= 0).sum()
)

```

```python

print("Delivery Status Counts:")

display(
    df_50k["Delivery Status"]
    .value_counts(dropna=False)
    .to_frame("Order Count")
)

```

```python
# Create a copy of the original dataset
df_clean = df_50k.copy()

# Remove exact duplicate rows
df_clean = df_clean.drop_duplicates()

print("Rows before cleaning:", len(df_50k))
print("Rows after removing duplicates:", len(df_clean))
print("Duplicates remaining:", df_clean.duplicated().sum())
```

```python
# Check cleaned dataset
print("Cleaned Dataset Shape:", df_clean.shape)

print("\nDuplicate Rows:")
print(df_clean.duplicated().sum())

print("\nMissing Values:")
print(df_clean.isnull().sum())

print("\nDelivery Status After Cleaning:")
display(df_clean["Delivery Status"].value_counts().to_frame("Order Count"))
```

```python
# Convert date columns to datetime format
date_columns = [
    "order date (DateOrders)",
    "shipping date (DateOrders)"
]

for col in date_columns:
    df_clean[col] = pd.to_datetime(
        df_clean[col],
        errors="coerce"
    )

# Check date column types
print("Date Column Data Types:")
print(df_clean[date_columns].dtypes)

# Check invalid dates
print("\nInvalid Dates:")
print(df_clean[date_columns].isnull().sum())

# Check order and shipping dates
print("\nDate Range:")
print("Earliest Order Date:",
      df_clean["order date (DateOrders)"].min())

print("Latest Order Date:",
      df_clean["order date (DateOrders)"].max())

print("Earliest Shipping Date:",
      df_clean["shipping date (DateOrders)"].min())

print("Latest Shipping Date:",
      df_clean["shipping date (DateOrders)"].max())

# Check shipping date before order date
invalid_sequence = (
    df_clean["shipping date (DateOrders)"]
    < df_clean["order date (DateOrders)"]
)

print("\nShipping Dates Before Order Dates:",
      invalid_sequence.sum())
```

```python
# Calculate shipping delay
df_clean["Shipping Delay"] = (
    df_clean["Days for shipping (real)"]
    - df_clean["Days for shipment (scheduled)"]
)

# Classify shipping performance
def classify_shipping(delay):
    if delay > 0:
        return "Delayed"
    elif delay < 0:
        return "Ahead of Schedule"
    else:
        return "On Schedule"

df_clean["Shipping Performance"] = (
    df_clean["Shipping Delay"].apply(classify_shipping)
)

# Display results
print("Shipping Performance Summary:")
display(
    df_clean["Shipping Performance"]
    .value_counts()
    .to_frame("Order Count")
)

print("\nAverage Shipping Delay:")
print(round(df_clean["Shipping Delay"].mean(), 2), "days")

print("\nSample Records:")
display(
    df_clean[
        [
            "Order Id",
            "Days for shipping (real)",
            "Days for shipment (scheduled)",
            "Shipping Delay",
            "Shipping Performance",
            "Delivery Status"
        ]
    ].head(10)
)
```

```python
# Export cleaned dataset
output_file = "Supply_Chain_Cleaned_Data.csv"

df_clean.to_csv(output_file, index=False)

print("File exported successfully!")
print("File name:", output_file)
print("Total rows:", df_clean.shape[0])
print("Total columns:", df_clean.shape[1])

print("\nFinal Columns:")
print(df_clean.columns.tolist())
```

```python
import pandas as pd
import sqlite3

# Load cleaned CSV
df = pd.read_csv("Supply_Chain_Cleaned_Data.csv")

# Create SQLite database
conn = sqlite3.connect("supply_chain.db")

# Import data into SQL table
df.to_sql(
    "supply_chain",
    conn,
    if_exists="replace",
    index=False
)

print("Data imported successfully!")
```

```python
query = """
SELECT COUNT(*) AS total_records
FROM supply_chain;
"""

result = pd.read_sql_query(query, conn)
display(result)
```

```python
query = """
SELECT
    "Shipping Performance",
    COUNT(*) AS total_orders
FROM supply_chain
GROUP BY "Shipping Performance"
ORDER BY total_orders DESC;
"""

result = pd.read_sql_query(query, conn)
display(result)
```

```python
query = """
SELECT
    "Shipping Performance",
    COUNT(*) AS total_orders,
    ROUND(
        COUNT(*) * 100.0 / (SELECT COUNT(*) FROM supply_chain),
        2
    ) AS percentage
FROM supply_chain
GROUP BY "Shipping Performance"
ORDER BY total_orders DESC;
"""

result = pd.read_sql_query(query, conn)
display(result)
```

```python
query = """
SELECT
    "Shipping Mode",
    COUNT(*) AS total_records,
    SUM(
        CASE
            WHEN "Shipping Performance" = 'Delayed'
            THEN 1
            ELSE 0
        END
    ) AS delayed_records,
    ROUND(
        SUM(
            CASE
                WHEN "Shipping Performance" = 'Delayed'
                THEN 1.0
                ELSE 0
            END
        ) * 100.0 / COUNT(*),
        2
    ) AS delay_percentage
FROM supply_chain
GROUP BY "Shipping Mode"
ORDER BY delayed_records DESC;
"""

result = pd.read_sql_query(query, conn)
display(result)
```

```python
query = """
SELECT
    "Category Name",
    COUNT(*) AS total_records,
    SUM(
        CASE
            WHEN "Shipping Performance" = 'Delayed'
            THEN 1
            ELSE 0
        END
    ) AS delayed_records,
    ROUND(
        SUM(
            CASE
                WHEN "Shipping Performance" = 'Delayed'
                THEN 1.0
                ELSE 0
            END
        ) * 100.0 / COUNT(*),
        2
    ) AS delay_percentage
FROM supply_chain
GROUP BY "Category Name"
ORDER BY delayed_records DESC
LIMIT 10;
"""

result = pd.read_sql_query(query, conn)
display(result)
```

```python
query = """
SELECT
    strftime('%Y-%m', "order date (DateOrders)") AS order_month,
    COUNT(*) AS total_records,
    SUM(
        CASE
            WHEN "Shipping Performance" = 'Delayed'
            THEN 1
            ELSE 0
        END
    ) AS delayed_records,
    ROUND(
        100.0 * SUM(
            CASE
                WHEN "Shipping Performance" = 'Delayed'
                THEN 1
                ELSE 0
            END
        ) / COUNT(*),
        2
    ) AS delay_percentage
FROM supply_chain
GROUP BY order_month
ORDER BY order_month;
"""

result = pd.read_sql_query(query, conn)
display(result)
```

```python
query = """
SELECT
    COUNT(*) AS total_records,

    SUM(
        CASE
            WHEN "Shipping Performance" = 'Delayed'
            THEN 1 ELSE 0
        END
    ) AS delayed_records,

    ROUND(
        100.0 * SUM(
            CASE
                WHEN "Shipping Performance" = 'Delayed'
                THEN 1 ELSE 0
            END
        ) / COUNT(*),
        2
    ) AS delay_percentage,

    ROUND(
        AVG("Days for shipping (real)"),
        2
    ) AS average_actual_shipping_days,

    ROUND(
        AVG("Days for shipment (scheduled)"),
        2
    ) AS average_scheduled_shipping_days,

    ROUND(
        AVG("Shipping Delay"),
        2
    ) AS average_shipping_delay

FROM supply_chain;
"""

result = pd.read_sql_query(query, conn)
display(result)
```
